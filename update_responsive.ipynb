```python
from pathlib import Path

root = Path('.')
html_files = sorted(root.glob('*.html'))
print(f'Found {len(html_files)} HTML files in {root.resolve()}')
print('First few HTML files:', [p.name for p in html_files[:10]])
```

## Add Next Code Cell

Inject `responsive.css` and `responsive.js` into each HTML file if missing.

```python
css_link = '<link rel="stylesheet" href="responsive.css">'
js_link = '<script src="responsive.js"></script>'
added_css = []
added_js = []

for path in html_files:
    text = path.read_text(encoding='utf8')
    updated = text
    if 'responsive.css' not in text and '</head>' in text:
        updated = updated.replace('</head>', f'    {css_link}\n</head>', 1)
        added_css.append(path.name)
    if 'responsive.js' not in updated and '</body>' in updated:
        updated = updated.replace('</body>', f'    {js_link}\n</body>', 1)
        added_js.append(path.name)
    if updated != text:
        path.write_text(updated, encoding='utf8')

print('Added CSS to', len(added_css), 'files')
print('Added JS to', len(added_js), 'files')
print('Sample CSS added:', added_css[:10])
print('Sample JS added:', added_js[:10])
```

## Validate with Tests

Check that `responsive.css` and `responsive.js` appear in HTML files.

```python
missing_css = []
missing_js = []

for path in html_files:
    text = path.read_text(encoding='utf8')
    if 'responsive.css' not in text:
        missing_css.append(path.name)
    if 'responsive.js' not in text:
        missing_js.append(path.name)

print('Missing CSS in', len(missing_css), 'files')
print('Missing JS in', len(missing_js), 'files')
print('Missing CSS sample:', missing_css[:10])
print('Missing JS sample:', missing_js[:10])
```

## Document Observations and Next Steps

- Confirm whether responsive assets have been injected.
- Identify any HTML files still missing the links.
- If needed, update `responsive.css` further with page-specific mobile tweaks.
